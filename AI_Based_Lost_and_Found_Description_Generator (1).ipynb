```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "google/flan-t5-small"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

print("AI model loaded successfully!")
```

```python
def generate_description(item, color, location, extra_details):
    prompt = f"""
    Create a clear and detailed lost and found item description.

    Item: {item}
    Color: {color}
    Location: {location}
    Additional details: {extra_details}

    Write one natural and useful description for reporting the item.
    """

    inputs = tokenizer(prompt, return_tensors="pt")

    outputs = model.generate(
        **inputs,
        max_new_tokens=80
    )

    description = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return description

print("Description generator is ready!")
```

```python
prompt = """
Write a detailed lost-and-found description for this item:

Item: black leather wallet
Location found: college library
Additional details: small logo on the wallet

The description must be 2 to 3 complete sentences.
Mention the item's appearance, location, and identifying features.
Do not use placeholders. Write only the final description.
"""

inputs = tokenizer(prompt, return_tensors="pt")

outputs = model.generate(
    **inputs,
    max_new_tokens=100,
    num_beams=5
)

result = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("Generated Description:")
print(result)
```

```python
import gradio as gr

def create_description(report_type, item, color, location, extra_details):

    report_type = report_type.strip()
    item = item.strip()
    color = color.strip()
    location = location.strip()
    extra_details = extra_details.strip()

    # Make the item description natural
    item_description = f"{color} {item}"

    # Professional Lost report
    if report_type == "Lost":

        description = (
            f"A {item_description.lower()} has been reported missing "
            f"from {location.lower()}. "
            f"It features {extra_details.lower()}, which can be used "
            f"as an identifying characteristic. "
            f"Anyone who locates the item is requested to report it "
            f"to the lost-and-found department."
        )

    # Professional Found report
    else:

        description = (
            f"A {item_description.lower()} was found at "
            f"{location.lower()}. "
            f"It features {extra_details.lower()}, which can be used "
            f"as an identifying characteristic. "
            f"The owner is requested to verify these details and "
            f"contact the lost-and-found department to claim the item."
        )

    return description


app = gr.Interface(
    fn=create_description,

    inputs=[
        gr.Dropdown(
            choices=["Lost", "Found"],
            label="Report Type",
            value="Lost"
        ),

        gr.Textbox(
            label="Item Name",
            placeholder="Example: Wallet"
        ),

        gr.Textbox(
            label="Color",
            placeholder="Example: Black"
        ),

        gr.Textbox(
            label="Location",
            placeholder="Example: College Library"
        ),

        gr.Textbox(
            label="Additional Details",
            placeholder="Example: Leather wallet with a small logo"
        )
    ],

    outputs=gr.Textbox(
        label="Professional Generated Description",
        lines=7
    ),

    title="AI-Based Lost and Found Description Generator",

    description=(
        "Enter the details of a lost or found item to generate "
        "a professional Lost and Found report."
    ),

    submit_btn="Generate Professional Description",
    clear_btn="Clear"
)

app.launch(share=True)
```
